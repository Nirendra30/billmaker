```python
#04.10.2026
#Mini Challenge: Indian Grocery Bill Generator
item = input("Item Name:")
qty = int(input("Quantity"))
price = float(input("Price Per Unit:"))
gst_rate = 5 #5% GST on Groceries
discount = 10 #flat 10 rupees off

amount = qty*price
gst = amount*gst_rate/100
total = amount+gst-discount
print("\n==============================================================")
print("                   STUDY DELIGHT GROCERY MART                  ")
print("  ==============================================================")
print(f"  Item          : {item}")
print(f"  Qauntity      : {qty}")
print(f" Price/Unit     : {price}")
print(f"----------------------------------------------------------------")
print(f" Amount         : {amount}")
print(f" GST (5%)       : {gst}")
print(f" Discount       : {discount}")
print("------------------------------------------------------------------")
print(f" TOTAL          : ₹{total}")
print("==================================================================")
```
